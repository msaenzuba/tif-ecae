```python
#Funcionando bien al parecer

from bs4 import BeautifulSoup
nombre_archivo = "migrantes rusas argentina 2023"
file_path = r"C:\Users\marti\Documents\Facu\Especialización\Trabajo de investigación final\Fuentes\Facebook\código_fuente_FB_migrantes rusas argentina 2023.html"

with open(file_path, 'r', encoding='utf-8') as file:
    html_content = file.read()
soup = BeautifulSoup(html_content, 'html.parser')

text = soup.get_text()
print(text)
output_file_path = f"{nombre_archivo}.txt"  # Specify the output file name
with open(output_file_path, 'w', encoding='utf-8') as output_file:
    output_file.write(text)

print(f"Extracted text has been saved to {output_file_path}")

import csv
import re
separator = "FacebookFacebookFacebookFacebookFacebookFacebook"
split_text = text.split(separator)

# Save the split text into a CSV file
output_csv_path = f"{nombre_archivo}.csv"  # Specify the output CSV file name
with open(output_csv_path, 'w', encoding='utf-8', newline='') as csv_file:
    writer = csv.writer(csv_file)
    
    # Write each part of the split text as a new row in the CSV
    for part in split_text:
        writer.writerow([part.strip()])  # Strip leading/trailing whitespace

print(f"Split text has been saved to {output_csv_path}")

import pandas as pd

# Replace 'input_file.csv' with the path to your CSV file
input_csv_path = f"{nombre_archivo}.csv"
output_limpio = f"{nombre_archivo} limpio.csv"

# Read the CSV file into a DataFrame
df = pd.read_csv(input_csv_path, header=None)  # Use `header=None` if the file has no headers
df = df.dropna(how='all') # Drop empty rows
df = df.drop_duplicates() 
df.to_csv(output_limpio, index=False, header=False)  # Set `header=False` if the original file has no headers

print(f"Cleaned CSV has been saved to {output_limpio}")
```
